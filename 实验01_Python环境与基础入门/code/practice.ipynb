## 3. 条件与缩进

条件判断的思路与 C 相同，Python 用冒号和缩进表示分支的代码块，`elif` 表示继续判断另一种情况。
运行后把温度分别改为 32、18，观察进入哪个分支。多个条件可以用 `and`、`or` 连接，取反用 `not`。


```python
# PRACTICE P3：依次试 25、32、18。
temperature = 25
if temperature >= 30:
    message = "偏热"
elif temperature >= 20:
    message = "适中"
else:
    message = "偏冷"
print(temperature, "度：", message)

is_raining = False
can_go_out = temperature >= 20 and not is_raining
print("满足示例中的外出条件：", can_go_out)
```

## 4. 循环：直接取元素，也可以按次数重复

在 C 中遍历数组通常使用下标。Python 的 `for value in values` 可以直接取出每个元素；需要下标时可以用 `range(len(values))`。
先运行观察两种写法的输出，再补全累加的一行。


```python
values = [18, 21, 24]
for value in values:
    print("直接取得：", value)
for index in range(len(values)):
    print("下标", index, "对应", values[index])

total = 0
for value in values:
    # PRACTICE P4a：将当前值累加进去，而不是覆盖前面的结果。
    total = value
check("循环累加", total, 63)
```

`while` 在条件成立时重复执行。下面已提供完整例子，只改变初始次数，观察循环何时结束。
循环中的更新使 `remaining` 逐渐接近退出条件，运行时保留这条更新。


```python
# PRACTICE P4b：将 3 改为 1 或 5。
remaining = 3
while remaining > 0:
    print("剩余次数：", remaining)
    remaining -= 1
print("循环结束")
```

## 6. 类：把数据和操作放在一起

下面给出完整的计数器类。`__init__` 设置新对象的数据，`self` 指向当前对象，`add` 方法修改这个对象的计数。
本节通过创建和使用对象观察关系，不重复实现书店的类定义。


```python
class Counter:
    def __init__(self, value=0):
        self.value = value

    def add(self, amount):
        self.value += amount


first = Counter()
second = Counter(10)
first.add(2)
# PRACTICE P6a：再写一次方法调用，让 first 的计数增加 3。

print("first：", first.value, "；second：", second.value)
check("第一个对象的计数", first.value, 5)
check("另一个对象没有被修改", second.value, 10)

alias = first
alias.add(1)
print("通过 alias 修改后，first：", first.value)
print("alias 和 first 是否为同一个对象：", alias is first)
```

### 继承与重写：保留计数功能，增加日志

`LoggedCounter` 继承 `Counter`，继续使用父类的初始化方法。它重写 `add`，先通过 `super()` 完成原来的更新，再打印日志。
先运行下面的定义，再把对象列表里的第二个 `Counter()` 改为 `LoggedCounter()`。循环中的方法调用不变，观察新增输出。
这与项目中同一读取流程调用不同子类的解析方法相呼应。


```python
class LoggedCounter(Counter):
    def add(self, amount):
        super().add(amount)
        print("更新后的计数：", self.value)


# PRACTICE P6b：将第二个对象改成 LoggedCounter()。
counters = [Counter(), Counter()]
for counter in counters:
    counter.add(2)
print("两个对象的计数：", [counter.value for counter in counters])
```
